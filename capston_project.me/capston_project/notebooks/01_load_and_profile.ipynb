```python
import numpy as np
from freshflow.loader import load_sales, save_array, COL

data = load_sales("../data/raw/freshflow_sales.csv")
```

```python
print("Shape:", data.shape)
print("Dimensions:", data.ndim)
print("Dtype:", data.dtype)
print("Total elements:", data.size)
print("Bytes per element:", data.itemsize)
print("Total bytes:", data.nbytes)
print(data[:5])
```

```python
for name, index in COL.items():
    column = data[:, index]
    print(f"{name:20} min={np.nanmin(column):10.2f}  max={np.nanmax(column):10.2f}  missing={np.isnan(column).sum()}")
```

```python
print("NaN values:", np.isnan(data).sum())
print("Infinite values:", np.isinf(data).sum())
```

```python
with open("../data/processed/blank_test.csv", "w") as file:
    file.write("id,a,b\n1,2.5,\n3,,4\n")

with open("../data/processed/blank_test.csv", "r") as file:
    test = np.genfromtxt(file, delimiter=",", skip_header=1, ndmin=2)

print(test)
```

```python
with open("../data/processed/clean_test.csv", "w") as file:
    file.write("id,a,b\n1,2.5,3\n4,5.5,6\n")

with open("../data/processed/clean_test.csv", "r") as file:
    clean = np.loadtxt(file, delimiter=",", skiprows=1)

print(clean)
```

```python
save_array("../data/processed/loaded_copy.csv", data)
copy = load_sales("../data/processed/loaded_copy.csv")
print("Same as original:", np.array_equal(data, copy, equal_nan=True))
```
